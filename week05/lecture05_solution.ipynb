```python
import pandas as pd
import plotly.express as px
import numpy as np

# Dataset: Airbnb London Listings

df = pd.read_csv('../data/airbnb_london.csv')
print(f"Loaded: {len(df)} listings")
print(df.describe().round(1))
```

```python
p95 = df['price'].quantile(0.95)
df_cap = df[df['price'] <= p95]
print(f"95th percentile price: £{p95:.0f}")
print(df_cap.groupby('room_type')['price'].describe().round(1))
```

## Task 1 — Histogram: price by room type (overlapping distributions)

**What to build:** A histogram showing price distributions for **Entire home/apt vs Private room** (exclude Shared room — too few observations) overlaid on the same chart.

**Requirements:**
- Both room types on the same chart (use `color='room_type'`)
- `barmode='overlay'` with `opacity=0.6` so both distributions are visible
- A vertical line for the median of EACH room type, differently coloured
- Insight title comparing the two distributions

> 💡 `df_cap[df_cap['room_type'].isin(['Entire home/apt','Private room'])]`


```python
# Task 1
# YOUR CODE HERE
import plotly.express as px
import plotly.graph_objects as go

# Exclude Shared room
df_filterd = df_cap[df_cap['room_type'].isin(['Entire home/apt','Private room'])]

fig = px.histogram(
    data_frame=df_filterd,
    x='price',
    color='room_type',
    barmode='overlay',
    opacity=0.6,
    color_discrete_map={
        'Entire home/apt': '#2E75B6',
        'Private room': '#70AD47'
    },
    labels={'price': 'Nightly Price (£)', 'room_type': 'Room Type'},
    height=500
)

# Add median line
median_entire = df_filterd[df_filterd['room_type'] == 'Entire home/apt']['price'].median()
median_private = df_filterd[df_filterd['room_type'] == 'Private room']['price'].median()
fig.add_vline(x=median_entire, line_dash='dash', line_color='#E63946', line_width=2,
              annotation=dict(
                  text=f'<b>Median: £{median_entire:.0f}',
                  font=dict(color='#E63946', size=12),
                  xanchor='left',
                  yanchor='top',
                  xshift=8,   # pixels to the right of the line
              ))
fig.add_vline(x=median_private, line_dash='dash', line_color='#E63946', line_width=2,
              annotation=dict(
                  text=f'<b>Median: £{median_private:.0f}',
                  font=dict(color='#E63946', size=12),
                  xanchor='left',
                  yanchor='top',
                  xshift=8,   # pixels to the right of the line
              ))

fig.update_layout(
    title='Entire homes command significantly higher median prices and show a wider price distribution than private rooms',
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    xaxis=dict(gridcolor='#EEEEEE', title='Nightly Price (£)'),
    yaxis=dict(showgrid=False, title='Count'),
    showlegend=True,
    margin=dict(l=60, r=40, t=65, b=40)
)

fig.show()
```

## Task 2 — Box plot: listing activity by borough

**What to build:** A **horizontal box plot** comparing listing activity (reviews per month) across London boroughs — reviews per month is a proxy for how frequently a listing is booked.

**Requirements:**
- Horizontal orientation (borough names are long)
- Sorted by median reviews per month (most active at top)
- Highlight the **two most active** boroughs in a different colour
- Outliers shown as individual points
- Insight title naming the two busiest boroughs

> 💡 Some listings have zero reviews — these are new or inactive listings. Filter them out with before plotting

```python
# Task 2
# YOUR CODE HERE
df_zero_remove = df[df['reviews_per_month'] > 0].copy()

median_reviews = df_zero_remove.groupby('neighbourhood')['reviews_per_month'].median().sort_values(ascending=False)

top2_boroughs = median_reviews.head(2).index.tolist()

color_map = {}
for b in df_zero_remove['neighbourhood'].unique():
    if b in top2_boroughs:
        color_map[b] = '#2E75B6'
    else:
        color_map[b] = '#AAAAAA' 

sorted_boroughs = median_reviews.index.tolist()
sorted_boroughs.reverse() 

fig = px.box(
    data_frame=df_zero_remove,
    x='reviews_per_month', 
    y='neighbourhood',
    color='neighbourhood',
    color_discrete_map=color_map,
    notched=True,               
    points='outliers',          
    labels={'reviews_per_month': 'Reviews per Month', 'neibourhood': ''},
    category_orders={'neibourhood': sorted_boroughs}, 
    height=800
)

fig.update_layout(
    title='The most active - median reviews per month stand out',
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    xaxis=dict(gridcolor='#EEEEEE', title='Nightly Price (£)'),
    yaxis=dict(showgrid=False),
    showlegend=False,
    margin=dict(l=60, r=40, t=55, b=40)
)
fig.show()
```
