```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python
highlight = 'China'

color_map = {c: 'crimson' if c == highlight else 'lightgrey' for c in df['Country'].unique()}

fig = px.line(df, x='Year', y='CO2_per_capita', color='Country',
              color_discrete_map=color_map,
              labels={'CO2_per_capita': 'CO2_per_capita(pc)', 'Year': ''},
              title="China's CO2 emission per capita 2000–2022 - biggest pollution per capita across world",
              template='plotly_white',
              height=400
         )

fig.update_traces(
    line=dict(width=1),
    showlegend=False 
)

fig.update_traces(
    line=dict(width=3),
    selector=dict(name=highlight)
)

highlight_label = df.loc[(df['Country'] == highlight) & (df['Year'] == df['Year'].max())]
others_label = df.loc[(df['Country'] == 'India') & (df['Year'] == df['Year'].max())]

fig.add_annotation(
    x=highlight_label['Year'].values[0], y=highlight_label['CO2_per_capita'].values[0],
    text=highlight, showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='crimson', size=12, weight='bold', family='Arial')
)

fig.add_annotation(
    x=others_label['Year'].values[0], y=others_label['CO2_per_capita'].values[0],
    text='Other Countries', showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='lightgrey', size=12, weight='bold', family='Arial')
)

fig.update_layout(
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False)
)


fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python
# Task — Multi-series line with highlight
# YOUR CODE HERE
df_asia = df[df['Region'] == 'Asia']

color_map = {c: 'crimson' if c == 'China' else 'lightgrey' for c in df_asia['Country'].unique()}

fig = px.line(df_asia, x='Year', y='CO2_Mt', color='Country',
               color_discrete_map=color_map,
               labels={'CO2_Mt': 'CO2 Emissions (Mt)', 'Year': ''},
               title="China's CO2 emissions tripled 2000-2022 - biggest pollution across Asian countries",
               template='plotly_white',
               height=400
               )

fig.update_traces(
    line=dict(width=1),
    showlegend=False 
)

fig.update_traces(line=dict(width=2.5), selector=dict(name='China'))

china_label = df_asia.loc[(df_asia['Country'] == 'China') & (df_asia['Year'] == df_asia['Year'].max())]
india_label = df_asia.loc[(df_asia['Country'] == 'India') & (df_asia['Year'] == df_asia['Year'].max())] 

fig.add_annotation(
    x=china_label['Year'].values[0], y=china_label['CO2_Mt'].values[0],
    text='China', showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='crimson', size=12, weight='bold', family='Arial')
)

fig.add_annotation(
    x=india_label['Year'].values[0], y=india_label['CO2_Mt'].values[0],
    text='Other Asian Countries', showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='lightgrey', size=12, weight='bold', family='Arial')
)

fig.update_layout(
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False)
)

fig.show()
```
